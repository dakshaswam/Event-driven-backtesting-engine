```python
import sys
sys.path.append("../src")

from collections import deque
from pathlib import Path
import importlib.util

# load events module directly from the src/events.py file to avoid import-name/package issues
_events_path = Path("../src/events.py").resolve()
_spec = importlib.util.spec_from_file_location("events_module", _events_path)
_events = importlib.util.module_from_spec(_spec)
_spec.loader.exec_module(_events)

MarketEvent = _events.MarketEvent
SignalEvent = _events.SignalEvent
OrderEvent = _events.OrderEvent
FillEvent = _events.FillEvent

queue = deque()

queue.append(MarketEvent())
queue.append(SignalEvent("AAPL", "2005-01-03", "BUY"))
queue.append(OrderEvent("AAPL", "2005-01-03", "BUY", 100))
queue.append(FillEvent("AAPL", "2005-01-04", "BUY", 100, 0.95, 0.9505, 0.50))

while queue:
    event = queue.popleft()
    print(type(event).__name__, event)
```
